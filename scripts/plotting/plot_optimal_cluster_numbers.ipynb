```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("../../results/sc_blood_optimal_numbers_of_clusters.tsv", sep="\t", index_col=0)
data["tissue"] = data.index
data = data.replace({"cd8+_tcells" : "CD8+ T-cell", "dc" : "Dendritic cell", "nk_cells" :"NK cell", "monocytes" : "Monocyte", "b_cells" : "B-cell", "cd4+_tcells" : "CD4+ T-cell"})
data
```

```python
ax = sns.barplot(data=data, y="tissue", x="num_clusters", color="orange", edgecolor="black")

label_fontsize = 16

ax.set_xlabel('Suggested $\ell$ from heuristic', fontsize=label_fontsize)
ax.set_ylabel('Cell type', fontsize=label_fontsize)
ax.tick_params(axis='x', labelsize=label_fontsize)
ax.tick_params(axis='y', labelsize=label_fontsize)
ax.set_xticks([0,2,4,6,8,10,12,14])
plt.savefig("sc_blood_optimal_numbers_of_clusters.pdf", bbox_inches="tight")
```
