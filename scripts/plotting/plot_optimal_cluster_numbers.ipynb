```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("../../results/optimal_cluster_numbers_from_heuristic.tsv", sep="\t", index_col=0)
data["tissue"] = data.index
data
```

```python
ax = sns.barplot(data=data, y="tissue", x="num_clusters", color="orange", edgecolor="black")

label_fontsize = 16

ax.set_xlabel('Suggested $\ell$ from heuristic', fontsize=label_fontsize)
ax.set_ylabel('Tissue', fontsize=label_fontsize)
ax.tick_params(axis='x', labelsize=label_fontsize)
plt.savefig("optimal_numbers_of_clusters.pdf", bbox_inches="tight")
```
